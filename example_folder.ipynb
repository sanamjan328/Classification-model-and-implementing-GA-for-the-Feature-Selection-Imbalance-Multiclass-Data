```python
import os
import time
extentions = ['.ipynb']
str = "/Users/sanamjankurban/Documents/Jupyter_projects/Data Analysis Examples/ML Courses/"

def get_file_list(root_dir, E):
    file_list = []
    counter = 1
    
    for p, d, f in os.walk(root_dir):
        for filename in f:
            
            if any(ext in filename for ext in E):
                fullpath = os.path.join(p, filename)
                file_name = os.path.join(filename)
                size = os.path.getsize(fullpath)
                time_file = os.path.getctime(fullpath)
                create_time = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime(time_file))
                item_list = (fullpath, filename, size, create_time)
                file_list.append(item_list)
                counter += 1

    return file_list
```

```python
def get_file_list():
    file_list = []
    delta = datetime.timedelta(days=30)
    now = datetime.datetime.now()
    
    for p, d, f in os.walk(str):
        for filename in f:
            
            if filename.endswith(".ipynb"):
                fullpath = os.path.join(p, filename)
                file_name = os.path.join(filename)
                size = os.path.getsize(fullpath)
                time_file = datetime.datetime.fromtimestamp(os.path.getctime(fullpath))
                if (now-delta < time_file):
                    create_time = time_file.strftime("%Y-%m-%d %H:%M:%S")
                    item_list = (fullpath, filename, size, create_time)
                    file_list.append(item_list)
    return file_list
```
